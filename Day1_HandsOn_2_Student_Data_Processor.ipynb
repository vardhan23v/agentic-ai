### Starter Dataset

A **dictionary** stores data as **key-value pairs**. Here each key is a student name and each value is their marks.

````python
# Class dataset stored as a dictionary
students = {
    "Asha": 82,
    "Rahul": 45,
    "Priya": 91,
    "Arun": 68,
    "Neha": 39
}
````

---

## Lists -- Store Marks in Order

A **list** stores multiple values in order. This simple list contains the marks from our class dataset. List positions start at `0`, so `student_marks[0]` is the first mark. A `for` loop can visit each list value one at a time.

````python
# A list of marks in the same order as the class dataset
student_marks = [82, 45, 91, 68, 39]

# Access one value by position
print("First mark:", student_marks[0])

# Iterate through every value in the list
for mark in student_marks:
    print("Mark:", mark)
````

**Expected Output:**

```
First mark: 82
Mark: 82
Mark: 45
Mark: 91
Mark: 68
Mark: 39
```

**Learner Activity:** Change the list and use a `for` loop to print only the marks that are `50` or above.

````python
# Learner Activity: print marks that are 50 or above
for mark in student_marks:
    if mark >= 50:
        print(mark)
````

**Expected Output:**

```
82
91
68
```

The loop checks every list value and prints only the values that meet the condition.

---

## `range()` -- Use Positions to Number Students

`range()` produces a sequence of numbers. It is useful when you need a position, such as a student's number, while processing a list. The stop value is not included, so `range(len(student_names))` visits positions `0` through `4`.

````python
# Get student names in list order
student_names = list(students.keys())

# Number each student using their list position
for position in range(len(student_names)):
    print(position + 1, "-", student_names[position])
````

**Expected Output:**

```
1 - Asha
2 - Rahul
3 - Priya
4 - Arun
5 - Neha
```

`position` starts at `0`, so adding `1` displays familiar student numbers from `1` to `5`.

---

## `while` Loop -- Validate a Mark

A `while` loop repeats as long as a condition is true. It is useful when you do not know how many attempts someone will need. Here, the loop keeps asking for a mark until the value is between `0` and `100`. A `for` loop is better when you already know which collection items to visit.

````python

# Keep asking until the mark is valid
student_mark = int(input("Enter a student mark (0-100): "))

while student_mark < 0 or student_mark > 100:
    print("Please enter a mark between 0 and 100.")
    student_mark = int(input("Enter a student mark (0-100): "))

print("Accepted mark:", student_mark)
````

**Expected Output (example):**

```
Enter a student mark (0-100): 120
Please enter a mark between 0 and 100.
Enter a student mark (0-100): 88
Accepted mark: 88
```

The number of repetitions depends on the values entered. Once the mark is valid, the condition becomes false and the loop stops.

## Step 1 -- Explore the Dataset

Before processing data it is good practice to look at what you have. Display the full dictionary and the number of students.

````python
# Display the complete dataset
print("Student Data:", students)

# Display the number of students
print("Total Students:", len(students))
````

**Expected Output:**

```
Student Data: {'Asha': 82, 'Rahul': 45, 'Priya': 91, 'Arun': 68, 'Neha': 39}
Total Students: 5
```

`len()` returns the number of key-value pairs in the dictionary.

---

## Step 2 -- Loop Through Students

A `for` loop lets you visit every item in a collection **one at a time** without writing repetitive code.

The `.items()` method returns each **key** (student name) and its **value** (marks) together.

````python
# Loop through the dictionary and print each student
for name, marks in students.items():
    print(name, "-", marks)
````

**Expected Output:**

```
Asha - 82
Rahul - 45
Priya - 91
Arun - 68
Neha - 39
```

- `name` receives the **key** (e.g., `"Asha"`).
- `marks` receives the **value** (e.g., `82`).
- The loop body runs once for each pair.

---

## Step 3 -- Identify Pass / Fail

Add a condition inside the loop to classify each student. We will also count how many passed and how many failed.

````python
# Counters for pass and fail
pass_count = 0
fail_count = 0

# Loop through each student and determine result
for name, marks in students.items():
    if marks >= 50:
        result = "Pass"
        pass_count = pass_count + 1
    else:
        result = "Fail"
        fail_count = fail_count + 1

    print(name, "-", marks, "-", result)

print()
print("Passed:", pass_count)
print("Failed:", fail_count)
````

**Expected Output:**

```
Asha - 82 - Pass
Rahul - 45 - Fail
Priya - 91 - Pass
Arun - 68 - Pass
Neha - 39 - Fail

Passed: 3
Failed: 2
```

---

## Step 4 -- Calculate Class Average

To find the average you need two things: the **total** of all marks and the **number** of students.

We will use a loop to add up all the marks. This helps you practice how loops accumulate values.

````python
# Calculate total marks using a loop
total = 0

for name, marks in students.items():
    total = total + marks

# Calculate average
num_students = len(students)
average = total / num_students

print("Total Marks:", total)
print("Number of Students:", num_students)
print("Class Average:", round(average, 2))
````

**Expected Output:**

```
Total Marks: 325
Number of Students: 5
Class Average: 65.0
```

- `total` starts at `0` and grows with each loop iteration.
- `round(average, 2)` limits the result to 2 decimal places.

---

## Step 5 -- Find the Highest Performer

One way to find the student with the highest marks is to track the best score as you loop through the data.

````python
# Find the highest-performing student
top_name = ""
top_marks = 0

for name, marks in students.items():
    if marks > top_marks:
        top_marks = marks
        top_name = name

print("Highest Performer:", top_name, "with", top_marks, "marks")
````

**Expected Output:**

```
Highest Performer: Priya with 91 marks
```

**How it works:**

- `top_marks` starts at `0`.
- Each time the loop finds marks higher than `top_marks`, it updates both `top_marks` and `top_name`.
- After the loop finishes, the variables hold the best student's information.

**Note:** Python also has a built-in `max()` function that can do this in one line:

```python
top_name = max(students, key=students.get)
```

`max()` finds the key with the highest value. The `key=students.get` part tells `max()` to compare the dictionary values (marks) instead of the keys (names). The loop approach above shows you what `max()` is doing behind the scenes.

## Complete Class Report

Now let us combine all the steps into one clean report.

````python
# ---------- Complete Class Performance Report ----------

# Counters and accumulators
total = 0
pass_count = 0
fail_count = 0
top_name = ""
top_marks = 0

# Header
print("===== Class Performance Report =====")
print(f"{'Student':<12} {'Marks':<10} {'Result'}")
print("-" * 35)

# Process each student
for name, marks in students.items():
    # Determine result
    if marks >= 50:
        result = "Pass"
        pass_count = pass_count + 1
    else:
        result = "Fail"
        fail_count = fail_count + 1

    # Accumulate total
    total = total + marks

    # Track highest performer
    if marks > top_marks:
        top_marks = marks
        top_name = name

    # Print row
    print(f"{name:<12} {marks:<10} {result}")

# Summary
average = total / len(students)

print("-" * 35)
print("Class Average      :", round(average, 2))
print("Students Passed    :", pass_count)
print("Students Failed    :", fail_count)
print("Highest Performer  :", top_name, "(", top_marks, "marks )")
print("====================================")
````

**Expected Output:**

```
===== Class Performance Report =====
Student      Marks      Result
-----------------------------------
Asha         82         Pass
Rahul        45         Fail
Priya        91         Pass
Arun         68         Pass
Neha         39         Fail
-----------------------------------
Class Average      : 65.0
Students Passed    : 3
Students Failed    : 2
Highest Performer  : Priya ( 91 marks )
====================================
```

## Final Challenge

Extend the class report with the following improvements:

1. **Add 3 more students** to the `students` dictionary.
2. **Calculate both the highest and lowest scores** and display the corresponding student names.
3. **Count distinctions** -- students who scored 75 or above.
4. **Produce a complete class summary** that includes:
   - Total students
   - Distinctions count
   - Pass count
   - Fail count
   - Highest scorer
   - Lowest scorer
   - Class average

Write your solution in the cell below.

````python
# --- Your solution here ---

students = {
    "Asha": 82,
    "Rahul": 45,
    "Priya": 91,
    "Arun": 68,
    "Neha": 39,
    "vardhan": 85,
    "charan": 62,
    "tushar": 85,
    "thrishul": 78,

}

# Total students
total_students = len(students)

# Calculate class average
class_average = sum(students.values()) / total_students

# Find highest and lowest scorers
highest_score = max(students.values())
lowest_score = min(students.values())

highest_scorer = [name for name, score in students.items()
                  if score == highest_score]

lowest_scorer = [name for name, score in students.items()
                 if score == lowest_score]

# Count distinctions, passes and fails
distinctions = sum(1 for score in students.values() if score >= 75)
passes = sum(1 for score in students.values() if 50 <= score < 75)
fails = sum(1 for score in students.values() if score < 50)

# Display class summary
print("===== CLASS SUMMARY =====")
print("Total Students:", total_students)
print("Distinctions:", distinctions)
print("Pass:", passes)
print("Fail:", fails)
print("Highest Scorer:", highest_scorer[0], "-", highest_score)
print("Lowest Scorer:", lowest_scorer[0], "-", lowest_score)
print("Class Average:", round(class_average, 2))
````
