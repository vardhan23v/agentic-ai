## What is Prompt Engineering?

- Prompt engineering refers to the process of designing effective input prompts for LLMs to elicit accurate, relevant, and useful outputs.

- Since LLMs are sensitive to how a prompt is phrased, small changes in wording or format can significantly affect the response.

---

## Types of Prompts

| Type                    | Description |
|-------------------------|-------------|
| **Zero-shot**           | Asking the model to perform a task without providing any examples. |
| **Few-shot**            | Providing a few examples of the task to guide the model. |
| **Instructional**       | Giving the model clear instructions or roles. |
| **Chain-of-thought**    | Asking the model to reason step-by-step. |
| **Role-based**          | Assigning the model a specific persona or responsibility. |

---
##  Prompting Structure

A good prompt often follows this structure:

### 1. Instruction (What to do)
> Be clear and direct. Tell the model what you expect.

**Example**:

```text
Summarize the following paragraph in one sentence.
```

### 2. Context
Add any information the model should know before answering.

**Example:**

```text
You are an assistant for a financial firm. Only use simple, factual language.
```

### 3. Input (The actual data)
Provide the main text or content for the model to process.

**Example:**

```text
The financial report shows a 15% growth in Q2, driven by increased retail activity...
```

### 4. Format or Constraints (Optional)
Define what format the output should be in: bullet points, JSON, tables, etc.

**Example:**

```text
Respond in bullet points.
```

## How to Create Effective Prompts

Creating effective prompts involves careful design, clarity, and intent. Here are key strategies:

### 1. **Be Specific**
- Clearly state what you want.
- Avoid ambiguity or vague instructions.

**Example:**
- Bad: "Tell me something about the world."
- Good: "Give me three interesting facts about world geography."



### 2. **Set the Role or Context**
- Use a system message to define who the assistant is.

**Example:**
```json
{"role": "system", "content": "You are a senior software engineer helping junior developers."}
```
### 3. **Use Instructions**
- Add verbs like “explain”, “summarize”, “list”, “compare”, “write”.

**Example:**

- "List 5 benefits of using TypeScript in front-end development."

### 4. **Give Examples (Few-shot Prompting)**
- Show the format or logic to be followed.

**Example:**
```json
[
  {"role": "user", "content": "Convert to passive voice: 'She writes a letter.'"},
  {"role": "assistant", "content": "A letter is written by her."},
  {"role": "user", "content": "Convert to passive voice: 'They eat mangoes.'"}
]
```
### 5. **Encourage Step-by-Step Reasoning**

- Add: “Think step by step”, or “Explain your reasoning”.

**Example:**
```json
{"role": "user", "content": "If a product costs $80 and is discounted by 25%, what is the final price? Think step by step."}
```
### 6. **Set Format Expectations**
- Mention the format you want: JSON, bullet points, table, etc.

**Example:**

- "List 3 popular programming languages and their main features in a table format."

# 1. Zero-shot Prompting

- In zero-shot prompting, we give the model no examples — just a direct instruction or query.

- This relies entirely on the model’s pretraining and general reasoning ability.

- Simple and works for general queries.

````python
response = client.chat.completions.create(
    model="llama3-8b-instruct",
    messages=[
        {"role": "user", "content": "List three benefits of regular exercise."}
    ],
    temperature=0.7, # Example value, you can adjust this
    top_p=0.9 # Example value, you can adjust this
)
print(response.choices[0].message.content)
````

# 2. Few-shot Prompting

* We provide the model with a few labeled examples before asking it to complete a new similar task.

* This guides the model's behavior more effectively than zero-shot.

- Helps the model follow a pattern (e.g., polite tone, specific format).

````python
few_shot_prompt = [
    {"role": "system", "content": "You are a helpful assistant that rewrites sentences to be more formal."},
    {"role": "user", "content": "make it formal: wanna grab lunch?"},
    {"role": "assistant", "content": "Would you like to have lunch together?"},
    {"role": "user", "content": "make it formal: can't come now"}
]

response = client.chat.completions.create(
    model="llama3-8b-instruct",
    messages=few_shot_prompt
)
print(response.choices[0].message.content)
````

# 3. Instructional Prompting

- Give very clear, structured instructions to the model.

- Ensures output is in a specific format like bullet points, JSON, or a short paragraph.

- Helps in guiding the model's tone and role

- Great for task-based assistance like recommendation, planning, support

````python
response = client.chat.completions.create(
    model="llama3-8b-instruct",
    messages=[
        {"role": "system", "content": "You are a helpful travel advisor."},
        {"role": "user", "content": "Give me a 3-day itinerary for Paris under a budget."}
    ]
)
print(response.choices[0].message.content)
````

# 4. Chain-of-Thought Prompting

- Ask the model to think step by step or explain its reasoning.

- Promotes logical reasoning

* Improves accuracy in logic or math-based tasks.

- Best for math, programming, decision-making tasks

````python
response = client.chat.completions.create(
    model="llama3-8b-instruct",
    messages=[
        {"role": "user", "content": "A train travels 60 km in 1.5 hours. What is its average speed? Think step by step."}
    ]
)
print(response.choices[0].message.content)
````

# 5. Role-based Prompting

- You assign a role or identity to the AI (e.g., "act like a doctor").

- Sets the tone, style, and context of the conversation

- Great for entertainment, storytelling, education

- Makes the model respond in a specific tone, expertise, or profession.

````python
response = client.chat.completions.create(
    model="llama3-8b-instruct",
    messages=[
        {"role": "system", "content": "You are a Shakespearean poet who answers everything in rhyme."},
        {"role": "user", "content": "Explain how photosynthesis works."}
    ]
)
print(response.choices[0].message.content)
````

# Prompt Engineering Tips

- **Be clear:** Avoid ambiguous instructions.

- **Be specific:** Define roles and objectives.

- **Test variations:** Small prompt changes can improve output.

- **Use system role:** Helps control model behavior better.

- **Include examples:** In few-shot prompts, examples set patterns.
