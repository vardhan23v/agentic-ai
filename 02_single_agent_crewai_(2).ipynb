# CrewAI - Single Agent

This notebook demonstrates how to use **CrewAI** with a single agent to summarize any AI-related topic.

##  What this does:
- Takes a user input topic
- Creates a single summarizer agent
- Uses LLM to generate a summary

 Run each cell step-by-step to see it in action.

## 🔧 Importing Required Modules

We begin by importing the core components of **CrewAI**:

- `LLM`: Defines the language model to be used.
- `Agent`: Represents the individual AI worker (e.g., summarizer).
- `Task`: Describes what the agent is supposed to do.
- `Crew`: Coordinates the agent(s) and task(s) into a workflow.

We also import `textwrap` to format long text outputs for better readability.


```python
from crewai import LLM, Agent, Task, Crew
import textwrap
```

##  Initializing the Language Model (LLM)

We create an instance of the `LLM` class to define which language model the agent should use.

###  Parameters used:
- `model="gpt-4.1-nano"`: Specifies the LLM version.
- `temperature=0.7`: Controls the creativity of the response. (Lower = more focused, Higher = more diverse)
- `base_url`: The custom API endpoint for the LLM provider.
- `api_key`: Your API key to authenticate with the LLM service.

> **Note:** Make sure your API key is valid and the endpoint is accessible from your environment.


```python
llm = LLM(
    model="gpt-4.1-nano",
    temperature=0.7,
    base_url="https://nexusapi.navigatelabs.ai",
    api_key="")
```

## Creating the Summarizer Agent

We define a single **Agent** who will act as an AI tutor, capable of summarizing topics into clean, exam-ready points.

### Agent Configuration:
- `role="Indexed Summarizer"`: Defines the title or identity of the agent.
- `goal`: Specifies the objective — here, to break down AI topics into structured, indexed summaries.
- `backstory`: Gives the agent some "personality" or specialization — in this case, someone skilled at simplifying complex AI ideas.
- `tools=[]`: No external tools are used by this agent.
- `llm=llm`: Assigns the previously created language model.
- `verbose=True`: Enables detailed logging of the agent’s actions.

> This agent is designed to mimic an educator explaining AI topics clearly and effectively.


```python

summarizer = Agent(
    role="Indexed Summarizer",
    goal="Convert AI topics into structured, numbered summaries for exams",
    backstory="An expert in explaining complex AI topics clearly with indexed points.",
    tools=[],
    llm=llm,
    verbose=True
)
```

## Enter a Topic

This line asks the user to type an **AI topic** they want summarized.

The entered topic will be passed to the agent to generate a summary.


```python
topic = input("Enter the AI topic you want summarized: ")
```

## Defining the Summary Task

We create a **Task** that tells the agent what to do.

### Task Details:
- `description`: Asks the agent to summarize the user-given topic in **5–7 clear, numbered points** (ideal for exam prep).
- `expected_output`: Ensures the output is well-structured using numbered points.
- `agent=summarizer`: Assigns this task to the summarizer agent we defined earlier.


```python
summary_task = Task(
    description=f"Summarize the topic '{topic}' in 5–7 concise, clearly numbered points for exam preparation.",
    expected_output="A clear, point-wise summary using numbering (1., 2., etc.) to enhance readability.",
    agent=summarizer
)
```

## 🚀 Running the Crew and Getting the Result

We now run the crew to perform the task and display the result.

### 🛠️ What happens here:
- `Crew(tasks=[summary_task])`: Creates a crew with just one task.
- `crew.kickoff()`: Starts the process — the agent begins working on the assigned task.
- `print()`: Displays the final **indexed summary** generated by the agent.


```python
crew = Crew(tasks=[summary_task])
result = await crew.kickoff_async()
print("\n📘 Indexed Summary:\n",result)
```
