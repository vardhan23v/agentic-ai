# Nexus API Integration

```python
from google.colab import userdata
api_key=userdata.get('api_key')
```

This cell retrieves the Nexus API key securely from Colab's user data. Make sure you have stored your API key under the name `api_key` in the Colab secrets manager.

```python
from google.colab import userdata
import openai

api_key = userdata.get("api_key")

client = openai.OpenAI(
    api_key=api_key,
    base_url="https://nexusapi.navigatelabs.ai"
)

response = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[
        {
            "role": "user",
            "content": "Hi"
        }
    ]
)

print(response.choices[0].message.content)
```

This cell initializes the OpenAI client and makes a sample API call to the chat completions endpoint using the specified model and a simple prompt. The response is then printed.
